<p><font color="#0E7490"><b>What this means:</b></font> <font color="#0E7490">Use <code>elif</code> when there are <b>more than two possible answers</b>. Python checks conditions from top to bottom and uses the first one that is true. Use <code>and</code> when <b>both</b> tests must pass and <code>or</code> when <b>either</b> test may pass. <h2><font color="#1D4ED8"><b>3. Repeat a task: for loops</b></font></h2>

<p><font color="#0E7490"><b>What this means:</b></font> <font color="#0E7490">A <code>for</code> loop repeats a block of code <b>once for each item</b>. You can loop through a list, such as <code>["heart", "liver"]</code>, or through a set of numbers using <code>range()</code>. For example, <code>range(1, 4)</code> gives <code>1, 2, 3</code>—it stops before 4.</font></p>
<p><font color="#A34F00"><b>Remember:</b></font> <font color="#A34F00">Indent the code you want to repeat. Example: <code>for organ in organs:</code> followed by an indented <code>print(organ)</code>.</font></p>

<h3><font color="#0E7490"><b>Worked example 1: Print each structure</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>aorta / liver / kidney</code></p>

```python
structures = ["aorta", "liver", "kidney"]
for structure in structures:
    print(structure)
```

<h2><font color="#1D4ED8"><b>4. Repeat until a condition changes: while loops</b></font></h2>

<p><font color="#0E7490"><b>What this means:</b></font> <font color="#0E7490">A <code>while</code> loop repeats code <b>as long as a condition is true</b>. For example, <code>while number &lt;= 3:</code> keeps running until <code>number</code> becomes greater than 3. Unlike a <code>for</code> loop, a <code>while</code> loop needs you to update the changing value.</font></p>
<p><font color="#A34F00"><b>Remember:</b></font> <font color="#A34F00">Always change the value inside the loop, for example <code>number = number + 1</code>, so the loop eventually stops.</font></p>

<h3><font color="#0E7490"><b>Worked example 1: Count to three</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>1 / 2 / 3</code></p>

```python
number = 1
while number <= 3:
    print(number)
    number = number + 1
```

<h3><font color="#0E7490"><b>Worked example 2: Process three batches</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>Batches left: 3 / Batches left: 2 / Batches left: 1</code></p>

```python
remaining_batches = 3
while remaining_batches > 0:
    print("Batches left:", remaining_batches)
    remaining_batches = remaining_batches - 1
```

<h2><font color="#1D4ED8"><b>7. Work with labeled data: dictionaries</b></font></h2>

<p><font color="#0E7490"><b>What this means:</b></font> <font color="#0E7490">A <b>dictionary</b> stores related facts using names called <b>keys</b>. For example, <code>{"type": "CT", "slices": 120}</code> connects an image type and slice count. Get a value using its key, such as <code>scan["type"]</code>. A list of dictionaries can represent a tiny research table—one dictionary per scan.</font></p>
<p><font color="#A34F00"><b>Remember:</b></font> <font color="#A34F00">Use square brackets around the key, for example <code>scan["slices"]</code>. A dictionary uses curly braces <code>{ }</code>.</font></p>

<h3><font color="#0E7490"><b>Worked example 1: Read one scan record</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>ID: scan_01 / Slices: 120</code></p>

```python
scan = {"id": "scan_01", "type": "CT", "slices": 120}
print("ID:", scan["id"])
print("Slices:", scan["slices"])
```

<h3><font color="#0E7490"><b>Worked example 2: Read a list of records</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>A 80 / B 120</code></p>

```python
scans = [
    {"id": "A", "slices": 80},
    {"id": "B", "slices": 120}
]
for scan in scans:
    print(scan["id"], scan["slices"])
```

<h2><font color="#1D4ED8"><b>8. OPTIONAL: Group related information with classes</b></font> &nbsp; (optional)</h2>

<p><font color="#0E7490"><b>What this means:</b></font> <font color="#0E7490">A <b>class</b> is a blueprint for creating an object that keeps data and actions together. For example, a <code>Sample</code> object can store its <code>name</code>. Inside the class, <code>__init__</code> sets the starting data, and <code>self</code> means the particular object. A function inside a class is called a <b>method</b>.</font></p>
<p><font color="#A34F00"><b>Remember:</b></font> <font color="#A34F00">Classes are a stretch topic. Focus on recognizing <code>class</code>, <code>self</code>, and <code>__init__</code>; you do not need to memorize them yet.</font></p>

<h3><font color="#0E7490"><b>Worked example 1: Create a sample object</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>A</code></p>

```python
class Sample:
    def __init__(self, name):
        self.name = name

sample = Sample("A")
print(sample.name)
```

<h3><font color="#0E7490"><b>Worked example 2: Give an object a simple action</b></font></h3>
<p><font color="#334155">Try it:</font> Run the next cell. <font color="#0E7490"><b>Expected output:</b></font> <code>True</code></p>

```python
class Scan:
    def __init__(self, slices):
        self.slices = slices

    def has_many_slices(self):
        return self.slices >= 100

scan = Scan(120)
print(scan.has_many_slices())
```

<h2><font color="#177345"><b>8. Solutions: Classes (optional)</b></font></h2>

<p><font color="#177345"><b>Solution 8.1 — Store one property</b></font> &nbsp; <font color="#0E7490"><b>Expected:</b></font> <code>red</code></p>

```python
# Solution 8.1: Store one property
class Flower:
    def __init__(self, color):
        self.color = color

flower = Flower("red")
print(flower.color)
```

<p><font color="#177345"><b>Solution 8.2 — Store two properties</b></font> &nbsp; <font color="#0E7490"><b>Expected:</b></font> <code>liver 150</code></p>

```python
# Solution 8.2: Store two properties
class Organ:
    def __init__(self, name, volume):
        self.name = name
        self.volume = volume

organ = Organ("liver", 150)
print(organ.name, organ.volume)
```

<p><font color="#177345"><b>Solution 8.3 — Add a method</b></font> &nbsp; <font color="#0E7490"><b>Expected:</b></font> <code>CT scan</code></p>

```python
# Solution 8.3: Add a method
class Scan:
    def __init__(self, image_type):
        self.image_type = image_type

    def describe(self):
        print(self.image_type, "scan")

scan = Scan("CT")
scan.describe()
```

<p><font color="#177345"><b>Solution 8.4 — Return a calculated value</b></font> &nbsp; <font color="#0E7490"><b>Expected:</b></font> <code>12</code></p>

```python
# Solution 8.4: Return a calculated value
class Count:
    def __init__(self, number):
        self.number = number

    def double(self):
        return self.number * 2

count = Count(6)
print(count.double())
```

<p><font color="#177345"><b>Solution 8.5 — Use several objects</b></font> &nbsp; <font color="#0E7490"><b>Expected:</b></font> <code>A True / B False</code></p>

```python
# Solution 8.5: Use several objects
class Sample:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def passed(self):
        return self.score >= 80

samples = [Sample("A", 90), Sample("B", 65)]
for sample in samples:
    print(sample.name, sample.passed())
```

<h3><font color="#1D4ED8"><b>Quick reference</b></font></h3>
<p><font color="#0E7490"><b>if / else</b></font> — choose one result; example: <code>if count &gt; 5:</code>.</p>
<p><font color="#0E7490"><b>elif / and / or</b></font> — test several choices; example: <code>if x &gt; 0 and x &lt; 10:</code>.</p>
<p><font color="#0E7490"><b>for</b></font> — repeat for each item; example: <code>for x in [1, 2, 3]:</code>.</p>
<p><font color="#0E7490"><b>while</b></font> — repeat while something stays true; example: <code>while x &lt; 3:</code> (remember to update <code>x</code>).</p>
<p><font color="#0E7490"><b>def</b></font> — create a function; example: <code>def double(x): return 2 * x</code>.</p>
<p><font color="#0E7490"><b>list</b></font> — store values; example: <code>[10, 20, 30]</code>.</p>
<p><font color="#0E7490"><b>dictionary</b></font> — store labeled facts; example: <code>{"type": "CT"}</code>.</p>
<p><font color="#0E7490"><b>class (optional)</b></font> — combine related data and methods; example: <code>Sample("A")</code>.</p>
