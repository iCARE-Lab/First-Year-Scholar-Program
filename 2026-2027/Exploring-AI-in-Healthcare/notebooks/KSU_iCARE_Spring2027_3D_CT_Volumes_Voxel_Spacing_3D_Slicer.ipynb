<h2><font color="#1D4ED8"><b>2. Understanding 3D <code>.shape</code>, Voxels, and Millimeter Voxel Spacing</b></font></h2>

<p><font color="#0E7490"><b>What this means (Intuitive Concept):</b></font> <font color="#334155">Think of a <b>2D CT slice</b> like a single slice of bread, and a <b>3D CT volume</b> like the <b>entire loaf of bread</b> stacked together!</font></p>

<table style="width:100%;border-collapse:collapse;margin-top:8px;margin-bottom:8px;">
<tr style="background:#eaf3ff">
  <th style="padding:8px;text-align:left"><font color="#1D4ED8"><b>Property</b></font></th>
  <th style="padding:8px;text-align:left"><font color="#1D4ED8"><b>Command</b></font></th>
  <th style="padding:8px;text-align:left"><font color="#1D4ED8"><b>What It Tells You</b></font></th>
</tr>
<tr>
  <td style="padding:7px"><b>1. 3D Grid Shape</b></td>
  <td style="padding:7px"><code>ct_vol.shape</code></td>
  <td style="padding:7px">Returns 3 numbers: <code>(Width_X, Height_Y, Depth_Z)</code>—how many voxel cubes wide, front-to-back, and head-to-toe the scan is.</td>
</tr>
<tr style="background:#f8fafc">
  <td style="padding:7px"><b>2. Voxel Spacing (mm)</b></td>
  <td style="padding:7px"><code>sx, sy, sz = ct_nii.header.get_zooms()[:3]</code></td>
  <td style="padding:7px">The real-world physical size in <b>millimeters (mm)</b> of 1 single voxel cube along the X, Y, and Z axes.</td>
</tr>
<tr>
  <td style="padding:7px"><b>3. Physical Organ Volume (mL)</b></td>
  <td style="padding:7px"><code>(num_voxels * sx * sy * sz) / 1000.0</code></td>
  <td style="padding:7px">Multiplies the number of segmented organ voxels by the cubic-millimeter volume of 1 voxel, then divides by <code>1,000</code> to get <b>milliliters (mL)</b>!</td>
</tr>
</table>

<p><font color="#334155">Try it:</font> Run the next cell to load Patient 1 from <b><code>AortaSeg24</code></b> (`AortaSeg24_0001.nii.gz`), inspect its 3D shape and millimeter voxel spacing, and calculate the true 3D volume of the patient's aorta in milliliters (`mL`).</p>


```python
import nibabel as nib
import numpy as np

# 1. Load the first 3D CT scan and Ground-Truth mask from AortaSeg24
aorta_img_dir = os.path.join(CASES_DIR, "AortaSeg24", "images")
aorta_gt_dir  = os.path.join(CASES_DIR, "AortaSeg24", "ground_truth")

first_case_name = sorted(os.listdir(aorta_img_dir))[0]
ct_path   = os.path.join(aorta_img_dir, first_case_name)
mask_path = os.path.join(aorta_gt_dir,  first_case_name)

ct_nii   = nib.load(ct_path)
mask_nii = nib.load(mask_path)

ct_vol   = ct_nii.get_fdata()
mask_vol = mask_nii.get_fdata()

# 2. Read the real-world millimeter voxel spacing (sx, sy, sz)
sx, sy, sz = ct_nii.header.get_zooms()[:3]
one_voxel_mm3 = sx * sy * sz

# 3. Calculate the physical 3D Aorta Volume in milliliters (mL)
aorta_voxel_count = int(np.sum(mask_vol > 0))
aorta_volume_ml   = (aorta_voxel_count * one_voxel_mm3) / 1000.0

print("Patient Scan Filename:                  ", first_case_name)
print("3D CT Volume .shape (X, Y, Z):          ", ct_vol.shape)
print(f"Voxel Spacing (mm):                      ({sx:.2f} mm, {sy:.2f} mm, {sz:.2f} mm)")
print(f"Physical Dimensions of Scan (mm):        ({ct_vol.shape[0]*sx:.0f} mm × {ct_vol.shape[1]*sy:.0f} mm × {ct_vol.shape[2]*sz:.0f} mm)")
print(f"Volume of 1 single voxel cube:           {one_voxel_mm3:.2f} mm³")
print(f"Number of segmented Aorta voxels:        {aorta_voxel_count} voxels")
print(f"Real-World 3D Aorta Volume:              {aorta_volume_ml:.1f} mL")
```

<h2><font color="#1D4ED8"><b>3. Viewing High-Resolution 3-Plane Cross-Sections (Axial, Coronal, Sagittal)</b></font></h2>

<p><font color="#0E7490"><b>What this means (Intuitive Concept):</b></font> <font color="#334155">When radiologists open a 3D CT scan in <b>3D Slicer</b>, the software automatically does three things so the anatomy looks crisp and natural:</font></p>
<ol>
  <li><font color="#0E7490"><b>Physical Aspect-Ratio &amp; GPU Interpolation:</b></font> <font color="#334155">Uses the millimeter voxel spacing <code>(sx, sy, sz)</code> and GPU 3D resampling so the chest is never squished, stretched, or pixelated!</font></li>
  <li><font color="#0E7490"><b>Clinical CT Soft-Tissue / Vascular Windowing:</b></font> <font color="#334155">Focuses the grayscale brightness scale between <code>-160 HU</code> and <code>+450 HU</code> so contrast-filled blood vessels, kidneys, liver, and spine show maximum detail.</font></li>
  <li><font color="#177345"><b>The 3 Anatomical Planes:</b></font>
    <ul>
      <li><font color="#DC2626"><b>Axial Plane (Top-Down View):</b></font> <font color="#334155">Horizontal cross-section looking from head toward feet.</font></li>
      <li><font color="#177345"><b>Coronal Plane (Front-to-Back View):</b></font> <font color="#334155">Looking straight at the patient from the front.</font></li>
      <li><font color="#A34F00"><b>Sagittal Plane (Side Profile View):</b></font> <font color="#334155">Looking from the side—shows the full arch and descending trunk of the aorta along the spine!</font></li>
    </ul>
  </li>
</ol>

<p><font color="#334155">Try it:</font> Run the next cell to use PyTorch on the GPU to reconstruct a high-resolution, physically proportioned 3D volume and display the <b>Axial</b>, <b>Coronal</b>, and <b>Sagittal</b> views side by side!</p>


```python
import torch.nn.functional as F
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

def prepare_high_res_3d_on_gpu(ct_np, mask_np, zooms_mm, base_res=256):
    """
    Uses PyTorch on GPU (if available) to resample a 3D CT volume and mask into
    high-resolution, physically proportioned millimeter space so slices look crisp
    and never pixelated or stretched!
    """
    phys_mm = np.array(ct_np.shape[:3], dtype=float) * np.array(zooms_mm[:3], dtype=float)
    max_mm  = phys_mm.max()
    target_shape = tuple(np.clip(np.round((phys_mm / max_mm) * base_res).astype(int), 128, 320))

    # Move 3D arrays to GPU as 5D tensors: (Batch=1, Channel=1, X, Y, Z)
    ct_t   = torch.tensor(ct_np, dtype=torch.float32, device=device).unsqueeze(0).unsqueeze(0)
    mask_t = torch.tensor(mask_np, dtype=torch.float32, device=device).unsqueeze(0).unsqueeze(0)

    # Smooth 3D trilinear interpolation for CT; crisp nearest-neighbor for multi-label Mask
    ct_hr   = F.interpolate(ct_t, size=target_shape, mode="trilinear", align_corners=False).squeeze().cpu().numpy()
    mask_hr = F.interpolate(mask_t, size=target_shape, mode="nearest").squeeze().cpu().numpy().astype(int)

    # Apply clinical CT vascular window (-160 HU to +450 HU)
    ct_hr = np.clip(ct_hr, -160.0, 450.0)
    return ct_hr, mask_hr

# Reconstruct high-resolution, physically proportioned 3D CT and Mask on GPU
ct_hr, mask_hr = prepare_high_res_3d_on_gpu(ct_vol, mask_vol, (sx, sy, sz), base_res=256)
print("High-Resolution Proportionate 3D Shape:", ct_hr.shape)

# Find the slice index along each plane where the aorta is largest
x_mid = int(np.argmax(np.sum(mask_hr > 0, axis=(1, 2))))  # Sagittal slice
y_mid = int(np.argmax(np.sum(mask_hr > 0, axis=(0, 2))))  # Coronal slice
z_mid = int(np.argmax(np.sum(mask_hr > 0, axis=(0, 1))))  # Axial slice

views = [
    ("1. Axial View (Top-Down)",        np.rot90(ct_hr[:, :, z_mid]), np.rot90(mask_hr[:, :, z_mid])),
    ("2. Coronal View (Front-to-Back)", np.rot90(ct_hr[:, y_mid, :]), np.rot90(mask_hr[:, y_mid, :])),
    ("3. Sagittal View (Side Profile)", np.rot90(ct_hr[x_mid, :, :]), np.rot90(mask_hr[x_mid, :, :]))
]

fig = plt.figure(figsize=(14, 9), dpi=140)
for col_i, (title_text, ct_slice, msk_slice) in enumerate(views):
    # Top Row: Crisp Windowed CT Slice
    ax1 = plt.subplot(2, 3, col_i + 1)
    ax1.imshow(ct_slice, cmap="gray", vmin=-160, vmax=450, interpolation="bilinear")
    ax1.set_title(title_text, fontsize=11, fontweight="bold")
    ax1.axis("off")

    # Bottom Row: CT Slice + Smooth Multi-Zone Overlay + Crisp Outer Boundary Contour
    ax2 = plt.subplot(2, 3, col_i + 4)
    ax2.imshow(ct_slice, cmap="gray", vmin=-160, vmax=450, interpolation="bilinear")
    ax2.imshow(np.ma.masked_where(msk_slice == 0, msk_slice), cmap="Spectral", alpha=0.65, interpolation="nearest")
    if np.any(msk_slice > 0):
        ax2.contour(msk_slice > 0, levels=[0.5], colors="lime", linewidths=1.2)
    ax2.set_title(f"{title_text.split('(')[0].strip()} + 23-Zone Aorta Overlay", fontsize=11, fontweight="bold")
    ax2.axis("off")

plt.tight_layout()
plt.show()
```

<h2><font color="#1D4ED8"><b>4. Comparing Our 3D Datasets &amp; Opening a Scan in 3D Slicer</b></font></h2>

<p><font color="#0E7490"><b>What this means (Intuitive Concept):</b></font> <font color="#334155">Now let's compare a patient scan across our 3D datasets (<b><code>AortaSeg24</code></b>, <b><code>ImageTBAD</code></b>, and <b><code>BTCV</code></b>—plus <b><code>AVT</code></b>) and download a paired 3D CT scan (<code>sample_CT_scan.nii.gz</code>) and 3D segmentation mask (<code>sample_Aorta_mask.nii.gz</code>) to open in <b>3D Slicer</b> (<a href="https://www.slicer.org/" target="_blank">https://www.slicer.org/</a>):</font></p>

<div style="background:#f0fdf4;padding:12px;border-radius:8px;border-left:5px solid #177345;margin-bottom:10px;">
  <p><font color="#334155"><b>How to view your downloaded files in 3D Slicer:</b><br>
  1. Drag <code>sample_CT_scan.nii.gz</code> into 3D Slicer and click <b>OK</b>.<br>
  2. Drag <code>sample_Aorta_mask.nii.gz</code> into 3D Slicer, change <i>Description</i> from Volume to <b>Segmentation</b>, and click <b>OK</b>.<br>
  3. Switch to the <b>Segmentations</b> module at the top and click <b>Show 3D</b> to rotate the 3D aorta!</font></p>
</div>


```python
# Automatically include all datasets that have input CT images in cases_10
active_datasets = [d for d in ["AortaSeg24", "AVT", "ImageTBAD", "BTCV"]
                   if os.path.exists(os.path.join(CASES_DIR, d, "images")) and len(os.listdir(os.path.join(CASES_DIR, d, "images"))) > 0]

num_ds = len(active_datasets)
plt.figure(figsize=(4.2 * num_ds, 7.8), dpi=140)

for idx, dname in enumerate(active_datasets):
    img_folder = os.path.join(CASES_DIR, dname, "images")
    gt_folder  = os.path.join(CASES_DIR, dname, "ground_truth")
    fname = sorted(os.listdir(img_folder))[0]

    nii_ct = nib.load(os.path.join(img_folder, fname))
    nii_gt = nib.load(os.path.join(gt_folder,  fname))
    vol_ct, vol_gt = prepare_high_res_3d_on_gpu(
        nii_ct.get_fdata(), nii_gt.get_fdata(), nii_ct.header.get_zooms()[:3], base_res=256
    )

    if dname == "BTCV":
        # Axial slice shows the liver, spleen, kidneys, stomach, and aorta together
        best_slice = int(np.argmax(np.sum(vol_gt > 0, axis=(0, 1))))
        ct_2d, gt_2d = np.rot90(vol_ct[:, :, best_slice]), np.rot90(vol_gt[:, :, best_slice])
        view_name = "Axial View"
    else:
        # Sagittal slice shows the long longitudinal curve of the aorta
        best_slice = int(np.argmax(np.sum(vol_gt > 0, axis=(1, 2))))
        ct_2d, gt_2d = np.rot90(vol_ct[best_slice, :, :]), np.rot90(vol_gt[best_slice, :, :])
        view_name = "Sagittal View"

    # Top Row: Input 3D CT Slice
    plt.subplot(2, num_ds, idx + 1)
    plt.imshow(ct_2d, cmap="gray", vmin=-160, vmax=450, interpolation="bilinear")
    plt.title(f"{dname}: Input CT\n({fname.replace('.nii.gz', '')} — {view_name})", fontsize=10, fontweight="bold")
    plt.axis("off")

    # Bottom Row: Ground-Truth Segmentation Overlay + Contour
    plt.subplot(2, num_ds, idx + num_ds + 1)
    plt.imshow(ct_2d, cmap="gray", vmin=-160, vmax=450, interpolation="bilinear")
    plt.imshow(np.ma.masked_where(gt_2d == 0, gt_2d), cmap="Spectral", alpha=0.65, interpolation="nearest")
    if np.any(gt_2d > 0):
        plt.contour(gt_2d > 0, levels=[0.5], colors="lime", linewidths=1.1)
    plt.title(f"{dname}: Ground Truth Mask\n({int(np.max(vol_gt))} anatomical label(s))", fontsize=10, fontweight="bold")
    plt.axis("off")

plt.tight_layout()
plt.show()

# Export a paired CT scan and Mask for 3D Slicer
shutil.copy(ct_path, "sample_CT_scan.nii.gz")
shutil.copy(mask_path, "sample_Aorta_mask.nii.gz")
print("Prepared 'sample_CT_scan.nii.gz' and 'sample_Aorta_mask.nii.gz' for 3D Slicer!")

try:
    from google.colab import files
    files.download("sample_CT_scan.nii.gz")
    files.download("sample_Aorta_mask.nii.gz")
except Exception:
    print("Tip: Click the Folder icon on the left sidebar of Colab to download both .nii.gz files.")
```

```python
print("=== SOLUTION 1: 3D Shape & Millimeter Voxel Spacing ===")
aorta_dir   = os.path.join(CASES_DIR, "AortaSeg24", "images")
aorta2_file = sorted(os.listdir(aorta_dir))[1]
aorta2_nii  = nib.load(os.path.join(aorta_dir, aorta2_file))
aorta2_vol  = aorta2_nii.get_fdata()

print("3D Volume Shape:", aorta2_vol.shape)
print("Voxel Spacing (mm):", aorta2_nii.header.get_zooms()[:3])

print("\n=== SOLUTION 2: Count Segmented 3D Voxels in ImageTBAD ===")
tbad_gt_dir = os.path.join(CASES_DIR, "ImageTBAD", "ground_truth")
tbad_file   = sorted(os.listdir(tbad_gt_dir))[0]
tbad_mask   = nib.load(os.path.join(tbad_gt_dir, tbad_file)).get_fdata()

voxel_count = int(np.sum(tbad_mask > 0))
print(f"Total segmented aorta voxels in {tbad_file}: {voxel_count}")

print("\n=== SOLUTION 3: High-Resolution Axial Slice of BTCV ===")
axial_slice = np.rot90(btcv_hr_ct[:, :, z_center])
axial_gt    = np.rot90(btcv_hr_gt[:, :, z_center])

plt.figure(figsize=(8, 4), dpi=130)
plt.subplot(1, 2, 1)
plt.imshow(axial_slice, cmap="gray", vmin=-160, vmax=450, interpolation="bilinear")
plt.title(f"Solution 3: BTCV Axial CT (z = {z_center})")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(axial_slice, cmap="gray", vmin=-160, vmax=450, interpolation="bilinear")
plt.imshow(np.ma.masked_where(axial_gt == 0, axial_gt), cmap="Spectral", alpha=0.65, interpolation="nearest")
plt.title("BTCV 13-Organ Overlay")
plt.axis("off")
plt.tight_layout()
plt.show()
```
