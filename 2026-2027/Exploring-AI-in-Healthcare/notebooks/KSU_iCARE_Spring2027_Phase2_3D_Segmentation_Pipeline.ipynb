<h2><font color="#1D4ED8"><b>2. Experiencing 3D AI Training: How a 3D U-Net Learns on a 3D Volume</b></font></h2>

<p><font color="#0E7490"><b>What this means (Intuitive Concept):</b></font> <font color="#334155">In Fall 2026, you trained a <b>2D U-Net</b> on flat 2D X-rays using 2D filters (<code>nn.Conv2d</code>). How does a <b>3D U-Net</b> work on a 3D CT volume?</font></p>

<ul>
  <li><font color="#334155">Instead of sliding a flat <code>3×3</code> square across a 2D slice, a <b>3D U-Net</b> slides a <b><code>3×3×3</code> 3D cube filter (<code>nn.Conv3d</code>)</b> through all three dimensions (Width, Height, and Depth) at the same time!</font></li>
  <li><font color="#334155">And instead of 4D tensors <code>(Batch, Channels, H, W)</code>, a 3D U-Net uses <b>5D Tensors</b>: <b><code>(Batch, Channels, X, Y, Z)</code></b>!</font></li>
</ul>

<p><font color="#334155">Before we inspect the full 23-zone and 13-organ segmentations from our lab's supercomputer models, let's train a compact <b>Mini 3D U-Net</b> right here in Colab on a 3D Aorta scan for <b>25 epochs</b> so you can watch the <b>3D Segmentation Loss decrease</b> and the <b>3D Dice Score rise</b> in real time!</font></p>



```python
# 1. Load Patient 1 from AortaSeg24 for our live 3D training demonstration
demo_ct_path = os.path.join(CASES_DIR, "AortaSeg24", "images", "AortaSeg24_0001.nii.gz")
demo_gt_path = os.path.join(CASES_DIR, "AortaSeg24", "ground_truth", "AortaSeg24_0001.nii.gz")

raw_ct = nib.load(demo_ct_path).get_fdata()
raw_gt = (nib.load(demo_gt_path).get_fdata() > 0).astype(np.float32)  # Binary whole-aorta target (0 or 1)

# Scale CT intensities between 0.0 and 1.0 and prepare 5D tensors (1, 1, 64, 64, 64) for fast GPU training
ct_norm = (np.clip(raw_ct, -200, 600) + 200.0) / 800.0
x_5d = torch.tensor(ct_norm, dtype=torch.float32, device=device).unsqueeze(0).unsqueeze(0)
y_5d = torch.tensor(raw_gt,  dtype=torch.float32, device=device).unsqueeze(0).unsqueeze(0)

# Ensure 64x64x64 size for the fast 5-second live training demo (even if native 512x512 files are loaded)
x_train_3d = F.interpolate(x_5d, size=(64, 64, 64), mode="trilinear", align_corners=False)
y_train_3d = F.interpolate(y_5d, size=(64, 64, 64), mode="nearest")

# 2. Define a simple, intuitive 3D U-Net using 3D Convolutions (nn.Conv3d)
class Mini3DUNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.enc1 = nn.Sequential(nn.Conv3d(1, 8, 3, padding=1), nn.ReLU(), nn.Conv3d(8, 8, 3, padding=1), nn.ReLU())
        self.pool = nn.MaxPool3d(2)
        self.bot  = nn.Sequential(nn.Conv3d(8, 16, 3, padding=1), nn.ReLU(), nn.Conv3d(16, 16, 3, padding=1), nn.ReLU())
        self.up   = nn.Upsample(scale_factor=2, mode="trilinear", align_corners=False)
        self.dec1 = nn.Sequential(nn.Conv3d(16 + 8, 8, 3, padding=1), nn.ReLU(), nn.Conv3d(8, 1, 1))

    def forward(self, x):
        e1 = self.enc1(x)
        b  = self.bot(self.pool(e1))
        d1 = self.dec1(torch.cat([self.up(b), e1], dim=1))
        return torch.sigmoid(d1)

torch.manual_seed(42)
mini_3d_unet = Mini3DUNet().to(device)
optimizer = torch.optim.Adam(mini_3d_unet.parameters(), lr=0.02)

loss_curve_3d, dice_curve_3d = [], []

print("Training Mini 3D U-Net on 3D CT Volume (Shape:", tuple(x_train_3d.shape), ")...")
for epoch in range(1, 26):
    optimizer.zero_grad()
    pred_3d = mini_3d_unet(x_train_3d)

    # 3D Dice Score & Combined 3D Loss
    inter = (pred_3d * y_train_3d).sum()
    dice  = (2.0 * inter + 1e-5) / (pred_3d.sum() + y_train_3d.sum() + 1e-5)
    loss  = F.binary_cross_entropy(pred_3d, y_train_3d) + (1.0 - dice)

    loss.backward()
    optimizer.step()

    loss_curve_3d.append(loss.item())
    dice_curve_3d.append(dice.item())
    if epoch == 1 or epoch % 5 == 0:
        print(f"Epoch {epoch:2d}/25 | 3D Loss: {loss.item():.4f} (Decreasing!) | 3D Dice Score: {dice.item():.4f}")

# Plot the 3D Training Curves
plt.figure(figsize=(10, 3.5))
plt.subplot(1, 2, 1)
plt.plot(range(1, 26), loss_curve_3d, marker="o", color="#DC2626")
plt.title("3D U-Net Training Loss (Decreasing)")
plt.xlabel("Epoch"); plt.ylabel("3D BCE + Dice Loss"); plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(range(1, 26), dice_curve_3d, marker="s", color="#177345")
plt.title("3D U-Net Aorta Overlap (Rising 3D Dice)")
plt.xlabel("Epoch"); plt.ylabel("3D Dice Score"); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()

```

<h2><font color="#1D4ED8"><b>3. Side-by-Side Comparison: Original CT vs. Original Segmentation vs. AI-Trained Segmentations</b></font></h2>

<p><font color="#0E7490"><b>What this means (Intuitive Concept):</b></font> <font color="#334155">Now let's examine the full multi-class 3D segmentations produced by our lab's trained research models on <b><code>AortaSeg24</code></b> (which segments the aorta into <b>23 distinct anatomical zones and branching arteries</b>!). Below, we display a <b>4-column comparison</b> across all three anatomical planes (<b>Axial</b>, <b>Coronal</b>, and <b>Sagittal</b>):</font></p>
<ol>
  <li><font color="#334155"><b>Column 1 — Original 3D CT Scan:</b> The raw grayscale patient scan.</font></li>
  <li><font color="#334155"><b>Column 2 — Original Segmentation (Ground Truth):</b> The 23 aortic zones labeled by medical experts.</font></li>
  <li><font color="#0E7490"><b>Column 3 — Baseline AI Segmentation (<code>baseline_E02</code>):</b> Standard 3D <code>nnU-Net</code> prediction.</font></li>
  <li><font color="#177345"><b>Column 4 — AnatoGraph AI Segmentation (<code>anatograph_E10</code>):</b> Our anatomy-guided 3D model prediction!</font></li>
</ol>

<p><font color="#334155">Try it:</font> Run the next cell to display the 3-plane × 4-column visual comparison of Original vs. AI-Trained segmentations!</p>



```python
def load_case_4way(dataset_name, case_index=0):
    """Loads the Original CT, Ground Truth, Baseline E02, and AnatoGraph E10 volumes for a patient."""
    dpath = os.path.join(CASES_DIR, dataset_name)
    fname = sorted(os.listdir(os.path.join(dpath, "images")))[case_index]

    ct_nii  = nib.load(os.path.join(dpath, "images", fname))
    gt_vol  = nib.load(os.path.join(dpath, "ground_truth", fname)).get_fdata().astype(int)
    e02_vol = nib.load(os.path.join(dpath, "baseline_E02", fname)).get_fdata().astype(int)
    e10_vol = nib.load(os.path.join(dpath, "anatograph_E10", fname)).get_fdata().astype(int)
    ct_vol  = np.clip(ct_nii.get_fdata(), -160, 450)
    zooms   = ct_nii.header.get_zooms()[:3]
    return fname, ct_vol, gt_vol, e02_vol, e10_vol, zooms

fname, ct_v, gt_v, e02_v, e10_v, zooms = load_case_4way("AortaSeg24", case_index=0)

# Find the best slice index for Sagittal (x), Coronal (y), and Axial (z)
x_idx = int(np.argmax(np.sum(gt_v > 0, axis=(1, 2))))
y_idx = int(np.argmax(np.sum(gt_v > 0, axis=(0, 2))))
z_idx = int(np.argmax(np.sum(gt_v > 0, axis=(0, 1))))

# Aspect ratios from physical millimeter voxel spacing so anatomy is never stretched!
sx, sy, sz = zooms
plane_specs = [
    ("Axial View",    ct_v[:, :, z_idx], gt_v[:, :, z_idx], e02_v[:, :, z_idx], e10_v[:, :, z_idx], sy / sx),
    ("Coronal View",  ct_v[:, y_idx, :], gt_v[:, y_idx, :], e02_v[:, y_idx, :], e10_v[:, y_idx, :], sz / sx),
    ("Sagittal View", ct_v[x_idx, :, :], gt_v[x_idx, :, :], e02_v[x_idx, :, :], e10_v[x_idx, :, :], sz / sy)
]

fig = plt.figure(figsize=(15, 12), dpi=130)
col_titles = ["1. Original CT Scan", "2. Original Segmentation (GT)", "3. Baseline AI (E02)", "4. AnatoGraph AI (E10)"]

for row_i, (plane_name, s_ct, s_gt, s_e02, s_e10, aspect_ratio) in enumerate(plane_specs):
    s_ct_rot = np.rot90(s_ct)
    masks_rot = [None, np.rot90(s_gt), np.rot90(s_e02), np.rot90(s_e10)]

    for col_i in range(4):
        ax = plt.subplot(3, 4, row_i * 4 + col_i + 1)
        ax.imshow(s_ct_rot, cmap="gray", vmin=-160, vmax=450, aspect=aspect_ratio)
        if col_i > 0:
            m = masks_rot[col_i]
            ax.imshow(np.ma.masked_where(m == 0, m), cmap="tab20", alpha=0.72, aspect=aspect_ratio, interpolation="nearest")
        ax.set_title(f"{plane_name}\n{col_titles[col_i]}", fontsize=10, fontweight="bold")
        ax.axis("off")

plt.suptitle(f"AortaSeg24 ({fname}) — Original vs. AI-Trained 23-Zone Segmentations", fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()

```

```python
import shutil

gallery_datasets = ["AortaSeg24", "ImageTBAD", "BTCV"]
plt.figure(figsize=(15, 11.5), dpi=130)

for row_i, dname in enumerate(gallery_datasets):
    fname, ct_v, gt_v, e02_v, e10_v, (sx, sy, sz) = load_case_4way(dname, case_index=0)

    if dname == "BTCV":
        # Axial view shows the liver, spleen, kidneys, stomach, and aorta together
        best_s = int(np.argmax(np.sum(gt_v > 0, axis=(0, 1))))
        s_ct, s_gt, s_e02, s_e10 = ct_v[:, :, best_s], gt_v[:, :, best_s], e02_v[:, :, best_s], e10_v[:, :, best_s]
        aspect_r = sy / sx
    else:
        # Sagittal view shows the longitudinal curve of the aorta & dissection compartments
        best_s = int(np.argmax(np.sum(gt_v > 0, axis=(1, 2))))
        s_ct, s_gt, s_e02, s_e10 = ct_v[best_s, :, :], gt_v[best_s, :, :], e02_v[best_s, :, :], e10_v[best_s, :, :]
        aspect_r = sz / sy

    s_ct_rot = np.rot90(s_ct)
    panels = [
        ("Original CT",            None),
        ("Original Seg (Doctor)",  np.rot90(s_gt)),
        ("Baseline AI (E02)",      np.rot90(s_e02)),
        ("AnatoGraph AI (E10)",    np.rot90(s_e10))
    ]

    for col_i, (label_txt, mask_2d) in enumerate(panels):
        ax = plt.subplot(3, 4, row_i * 4 + col_i + 1)
        ax.imshow(s_ct_rot, cmap="gray", vmin=-160, vmax=450, aspect=aspect_r)
        if mask_2d is not None:
            ax.imshow(np.ma.masked_where(mask_2d == 0, mask_2d), cmap="tab20", alpha=0.72, aspect=aspect_r, interpolation="nearest")
        ax.set_title(f"{dname}: {label_txt}", fontsize=10, fontweight="bold")
        ax.axis("off")

plt.tight_layout()
plt.show()

# Export the AnatoGraph AI segmentation mask so students can load it into 3D Slicer!
ai_mask_src = os.path.join(CASES_DIR, "AortaSeg24", "anatograph_E10", "AortaSeg24_0001.nii.gz")
shutil.copy(ai_mask_src, "sample_AnatoGraph_AI_mask.nii.gz")
print("Prepared 'sample_AnatoGraph_AI_mask.nii.gz' to open in 3D Slicer alongside your CT scan!")

try:
    from google.colab import files
    files.download("sample_AnatoGraph_AI_mask.nii.gz")
except Exception:
    pass

```
