<h2><font color="#1D4ED8"><b>3. Data types: What kind of information is this?</b></font></h2>

Python needs to know what each value is. There are four common beginner types:

| Type | Meaning | Example |
|---|---|---|
| `str` | Text (string) | `"aorta"` |
| `int` | Whole number (integer) | `12` |
| `float` | Decimal number | `12.5` |
| `bool` | True or False | `True` |

A text value such as `"12"` is <font color="#0E7490"><b>not</b></font> the same as the number `12`.

<h3><font color="#0E7490"><b>3.1 Check a type with type()</b></font></h3>

`type()` tells you what kind of value you have.

```python
organ = "aorta"
case_count = 8
diameter_mm = 24.5
has_reference = True

print(type(organ))
print(type(case_count))
print(type(diameter_mm))
print(type(has_reference))
```

<h3><font color="#0E7490"><b>3.2 True and False</b></font></h3>

`True` and `False` are boolean values. Do not put quotation marks around them.

```python
has_reference_mask = True
is_complete = False
print(has_reference_mask)
print(type(is_complete))
```

<h3><font color="#0E7490"><b>3.3 Convert a value into another type</b></font></h3>

Use `int()`, `float()`, or `str()` when needed. For example, numbers that arrive as text may need conversion before calculations.

```python
count_as_text = "12"
count_as_number = int(count_as_text)
print(count_as_number + 3)

spacing_as_text = "1.5"
spacing_mm = float(spacing_as_text)
print(spacing_mm + 0.5)

case_number = 7
print("Case " + str(case_number))
```

<p><font color="#177345">Only convert text to a number if it really represents a number: <code>int("aorta")</code> will cause an error.</font></p>

<h2><font color="#1D4ED8"><b>4. Basic arithmetic: Working with numbers</b></font></h2>

Python works like a calculator.

| Symbol | Meaning | Example | Result |
|---|---|---|---|
| `+` | Add | `2 + 3` | `5` |
| `-` | Subtract | `7 - 2` | `5` |
| `*` | Multiply | `3 * 4` | `12` |
| `/` | Divide | `8 / 2` | `4.0` |
| `**` | Power | `2 ** 3` | `8` |
| `//` | Whole-number division | `7 // 2` | `3` |
| `%` | Remainder | `7 % 2` | `1` |

<font color="#0E7490"><b>Important:</b></font> Python uses `*` for multiplication, not `×`. Parentheses control the order of calculations.

<h3><font color="#0E7490"><b>4.1 Calculations</b></font></h3>

Run each line and look at the results. Division (`/`) returns a decimal result.

```python
print(8 + 4)
print(8 - 4)
print(8 * 4)
print(8 / 4)
print(2 ** 3)
```

<h3><font color="#0E7490"><b>4.2 Calculations using variables</b></font></h3>

Research data often consist of numbers that we want to combine.

```python
training_scans = 6
test_scans = 2
total_scans = training_scans + test_scans
print("Total scans:", total_scans)

diameter_mm = 24.0
radius_mm = diameter_mm / 2
print("Radius (mm):", radius_mm)
```

<h3><font color="#0E7490"><b>4.3 Parentheses and rounding</b></font></h3>

Use parentheses to make the intended calculation clear. `round(value, 2)` keeps two decimal places for display/calculations.

```python
print((3 + 5) * 2)
print(3 + 5 * 2)
print(round(8 / 3, 2))
```

<h3><font color="#0E7490"><b>5.2 Put variables inside sentences</b></font></h3>

<h3><font color="#0E7490"><b>8.1 Simple voxel-volume calculation</b></font></h3>

If each voxel measures 0.5 mm × 0.5 mm × 1.0 mm, its volume is the product of those three lengths. If 4,000 voxels belong to a fictional structure, multiply by that count.

```python
spacing_x_mm = 0.5
spacing_y_mm = 0.5
spacing_z_mm = 1.0
voxel_count = 4000

voxel_volume_mm3 = spacing_x_mm * spacing_y_mm * spacing_z_mm
structure_volume_mm3 = voxel_count * voxel_volume_mm3
print("Voxel volume (mm^3):", voxel_volume_mm3)
print("Structure volume (mm^3):", structure_volume_mm3)
print("Structure volume (mL):", structure_volume_mm3 / 1000)
```

<h3><font color="#0E7490"><b>8.2 A very simple comparison</b></font></h3>

A researcher may compare an AI-generated measurement against a reference measurement. `abs()` gives the size of a difference without a minus sign.

```python
reference_mm = 25
ai_mm = 27
difference_mm = abs(ai_mm - reference_mm)
print("Absolute difference (mm):", difference_mm)
```
